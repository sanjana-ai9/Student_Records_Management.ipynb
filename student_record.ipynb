```python
students = {
    101: {"name": "Ravi", "marks": 85},
    102: {"name": "Priya", "marks": 90},
    103: {"name": "Arun", "marks": 78}
}

print(students)
```

```python
def search_student(student_id):
    if student_id in students:
        print(students[student_id])
    else:
        print("Student not found")

search_student(102)
```

```python
def add_student(student_id, name, marks):
    students[student_id] = {"name": name, "marks": marks}
    print("Student added successfully")

add_student(104, "Sneha", 88)

print(students)
```

```python
def update_student(student_id, marks):
    if student_id in students:
        students[student_id]["marks"] = marks
        print("Student record updated successfully")
    else:
        print("Student not found")

update_student(102, 95)

print(students[102])
```

```python
def display_students():
    for student_id, details in students.items():
        print("Student ID:", student_id)
        print("Name:", details["name"])
        print("Marks:", details["marks"])
        print("------------------")

display_students()
```
